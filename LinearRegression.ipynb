```python
import pandas as pd
import statsmodels.api as sm
```

# Multiple linear Regression with all variables

```python
# Charger les données CSV
df = pd.read_csv("financial_data_completes.csv")

# Définir X et y
X = df.drop(columns=["Food CPI", "Month"])   
X = sm.add_constant(X)       
y = df["Food CPI"]

# Créer le modèle OLS (Ordinary Least Squares)
model = sm.OLS(y, X).fit()

# Résumé complet
print(model.summary())
```

# Drop  umsignificant variables

```python
X = X.drop(columns=["TRI_norm", "FAO Food Price Index", "FAO Meat", "FAO Dairy Index", "FAO Cereals Index", "FAO Oils Index", "FAO Sugar Index", "UCSB CHIRPS Rainfall: avg", "Palm oil ($/mt)", "Soybean oil ($/mt)", "Sunflower oil ($/mt)", "Maize ($/mt)", "Sorghum ($/mt)","Rice, Thai 25%  ($/mt)", "Rice, Viet Namese 5% ($/mt)", "Wheat, US SRW ($/mt)", "Wheat, US HRW ($/mt)", "Chicken ** ($/kg)", "Lamb ** ($/kg)", "Sugar, EU ($/kg)", "Sugar, world ($/kg)", "EUR/USD", "Crude oil, Brent ($/bbl)"]) 
```

```python
# Créer le modèle OLS (Ordinary Least Squares)
model = sm.OLS(y, X).fit()

# Résumé complet
print(model.summary())
```
